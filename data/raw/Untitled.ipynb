```python
import pandas as pd

path = r'C:\Users\ridas\Documents\air-quality-project\data\raw\marylebone_road_2025-2026.csv'
df = pd.read_csv(path, skiprows=10)
df = df.iloc[:-1]  # drop the "End" footer row

# numeric conversion, then interpolate gaps of up to 3 hours only
for c in ['Nitrogen dioxide',
          'PM10 particulate matter (Hourly measured)',
          'PM2.5 particulate matter (Hourly measured)']:
    df[c] = pd.to_numeric(df[c], errors='coerce')
    df[c] = df[c].interpolate(limit=3, limit_area='inside')

# datetime, including the 24:00:00 fix
df['datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'], errors='coerce')
mask = df['Time'] == '24:00:00'
df.loc[mask, 'datetime'] = pd.to_datetime(df.loc[mask, 'Date']) + pd.Timedelta(days=1)

df = df.set_index('datetime').drop(columns=['Date', 'Time']).sort_index()

print(df.shape)
print(df.index.isnull().sum())
```

```python
import numpy as np

no2 = 'Nitrogen dioxide'

# time features
df['hour'] = df.index.hour
df['dayofweek'] = df.index.dayofweek
df['month'] = df.index.month
df['is_weekend'] = df['dayofweek'].isin([5, 6]).astype(int)
df['is_rush_hour'] = df['hour'].isin([7, 8, 9, 17, 18, 19]).astype(int)

# cyclical encoding
df['hour_sin'] = np.sin(2 * np.pi * df['hour'] / 24)
df['hour_cos'] = np.cos(2 * np.pi * df['hour'] / 24)
df['month_sin'] = np.sin(2 * np.pi * df['month'] / 12)
df['month_cos'] = np.cos(2 * np.pi * df['month'] / 12)

# lag and rolling features (shift first so the current hour never leaks in)
df['no2_lag1'] = df[no2].shift(1)
df['no2_lag24'] = df[no2].shift(24)
df['no2_roll6'] = df[no2].shift(1).rolling(6).mean()
df['no2_roll24'] = df[no2].shift(1).rolling(24).mean()
```

```python
features = ['hour_sin', 'hour_cos', 'month_sin', 'month_cos',
            'is_weekend', 'is_rush_hour', 'dayofweek',
            'no2_lag1', 'no2_lag24', 'no2_roll6', 'no2_roll24']

model_df = df[features + [no2]].dropna()
print("Rows:", model_df.shape[0])

X = model_df[features]
y = model_df[no2]

split = int(len(model_df) * 0.8)
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]

print("Train:", X_train.index.min(), "to", X_train.index.max())
print("Test: ", X_test.index.min(), "to", X_test.index.max())
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

def report(name, pred):
    mae = mean_absolute_error(y_test, pred)
    rmse = np.sqrt(mean_squared_error(y_test, pred))
    print(f"{name}: MAE={mae:.2f}, RMSE={rmse:.2f}")

report("Training mean", np.full(len(y_test), y_train.mean()))
report("Persistence (same as 1h ago)", X_test['no2_lag1'])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

models = {
    'Linear Regression': LinearRegression(),
    'Random Forest': RandomForestRegressor(n_estimators=200, min_samples_leaf=5,
                                           n_jobs=-1, random_state=42),
    'XGBoost': XGBRegressor(n_estimators=300, learning_rate=0.05,
                            max_depth=4, random_state=42),
}

rows = {
    'Training mean (baseline)': {
        'MAE': mean_absolute_error(y_test, np.full(len(y_test), y_train.mean())),
        'RMSE': np.sqrt(mean_squared_error(y_test, np.full(len(y_test), y_train.mean())))},
    'Persistence (baseline)': {
        'MAE': mean_absolute_error(y_test, X_test['no2_lag1']),
        'RMSE': np.sqrt(mean_squared_error(y_test, X_test['no2_lag1']))},
}

preds = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    preds[name] = model.predict(X_test)
    rows[name] = {'MAE': mean_absolute_error(y_test, preds[name]),
                  'RMSE': np.sqrt(mean_squared_error(y_test, preds[name]))}

results = pd.DataFrame(rows).T.round(2)
results
```

```python
import matplotlib.pyplot as plt

imp = pd.Series(models['Random Forest'].feature_importances_, index=features).sort_values()
imp.plot(kind='barh', title='Feature importance (Random Forest)')
plt.tight_layout()
plt.show()
```

```python
best = 'Random Forest'  # change to whichever scored best
compare = pd.DataFrame({'actual': y_test, 'predicted': preds[best]}, index=y_test.index)
compare.iloc[-336:].plot(figsize=(14, 4), title=f'NO2: actual vs predicted ({best}), last 14 days')
plt.show()
```
